```python
import csv
import os

FILE = "expenses.csv"

def init_file():
    """Initialize the CSV file with headers if it does not exist."""
    if not os.path.exists(FILE):
        with open(FILE, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(["date", "category", "amount", "note"])

def get_amount():
    """Safely prompt the user for a numeric expense amount."""
    while True:
        raw = input("Amount: ")
        try:
            return float(raw)
        except ValueError:
            print("Please enter a valid number, e.g., 199.50")

def add_expense(date, category, amount, note=""):
    """Append a new expense record to the CSV file."""
    with open(FILE, "a", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([date, category, amount, note])
    print("Expense added successfully.")

def view_expenses():
    """Display all recorded expenses and calculate the total spent."""
    if not os.path.exists(FILE):
        print("No expenses recorded yet.")
        return

    total = 0
    print("\n--- All Expenses ---")
    try:
        with open(FILE, "r", newline="") as f:
            reader = csv.reader(f)
            header = next(reader, None)
            if not header:
                print("Expense log is empty.")
                return

            rows_found = False
            for row in reader:
                if len(row) < 4:
                    continue
                rows_found = True
                date, category, amount, note = row
                amount = float(amount)
                total += amount
                print(f"{date} | {category:10} | ₹{amount:.2f} | {note}")

            if not rows_found:
                print("No expenses found.")
            else:
                print(f"----------------------------------------")
                print(f"Total spent: ₹{total:.2f}")
    except FileNotFoundError:
        print("Expense file not found.")

def category_summary():
    """Display a category-wise breakdown of total spending."""
    if not os.path.exists(FILE):
        print("No expenses recorded yet.")
        return

    totals = {}
    try:
        with open(FILE, "r", newline="") as f:
            reader = csv.reader(f)
            header = next(reader, None)
            if not header:
                print("Expense log is empty.")
                return

            for row in reader:
                if len(row) < 4:
                    continue
                _, category, amount, _ = row
                totals[category] = totals.get(category, 0.0) + float(amount)

        if not totals:
            print("No category data available.")
            return

        print("\n--- Category Summary ---")
        for cat, amt in sorted(totals.items(), key=lambda x: -x[1]):
            print(f"{cat:10}: ₹{amt:.2f}")
        print("-" * 24)
    except FileNotFoundError:
        print("Expense file not found.")

def main():
    init_file()
    while True:
        print("\n=== Expense Tracker ===")
        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. Category Summary")
        print("4. Exit")

        choice = input("Enter choice (1-4): ").strip()

        if choice == "1":
            date = input("Date (YYYY-MM-DD): ").strip()
            category = input("Category: ").strip()
            amount = get_amount()
            note = input("Note (optional): ").strip()
            add_expense(date, category, amount, note)
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            category_summary()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 4.")

if __name__ == "__main__":
    main()
```
